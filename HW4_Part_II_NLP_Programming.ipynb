## Q1. NLP Preprocessing Pipeline

**Task:** Tokenize the sentence, remove English stopwords and punctuation, and apply stemming.

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
def preprocess(sentence):
    # Step 1: Tokenize the sentence
    tokens = word_tokenize(sentence)

    # Step 2: Remove stopwords and punctuation
    stop_words = set(stopwords.words('english'))
    filtered_tokens = [
        word for word in tokens
        if word.lower() not in stop_words and word.isalpha()
    ]

    # Step 3: Apply stemming
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

    print("Original Tokens:", tokens)
    print("Tokens Without Stopwords:", filtered_tokens)
    print("Stemmed Words:", stemmed_words)


sentence = "NLP techniques are used in virtual assistants like Alexa and Siri."
preprocess(sentence)
```

### Q1 – Short Answers

**1. What is the difference between stemming and lemmatization? Give an example with “running.”**

Stemming removes word endings to produce a stem, which may not be a valid dictionary word. Lemmatization uses vocabulary and grammatical information to produce a meaningful dictionary base form. For example, both can reduce “running” to “run,” but stemming may produce less meaningful forms for other words.

**2. Why remove stopwords, and when can removal be harmful?**

Removing stopwords can reduce text size and help focus on informative words in tasks such as keyword extraction. However, it can be harmful in sentiment analysis, translation, and question understanding because words such as “not” and “never” can change the meaning of a sentence.

## Q3. Scaled Dot-Product Attention

**Task:** Calculate the attention scores, scale them, apply softmax, and multiply the attention weights by V.

```python
import numpy as np

Q = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

K = np.array([[1, 0, 1, 0],
              [0, 1, 0, 1]])

V = np.array([[1, 2, 3, 4],
              [5, 6, 7, 8]])
```

```python
def scaled_dot_product_attention(Q, K, V):
    # Step 1: Dot product of Q and K transpose
    scores = np.dot(Q, K.T)

    # Step 2: Scale by the square root of the key dimension
    d_k = K.shape[1]
    scaled_scores = scores / np.sqrt(d_k)

    # Step 3: Apply softmax to each row
    exp_scores = np.exp(scaled_scores)
    attention_weights = exp_scores / np.sum(
        exp_scores, axis=1, keepdims=True
    )

    # Step 4: Multiply attention weights by V
    output = np.dot(attention_weights, V)

    return attention_weights, output


attention_weights, output = scaled_dot_product_attention(Q, K, V)

print("Attention Weights:")
print(attention_weights)

print("\nFinal Output:")
print(output)
```

### Q3 – Short Answers

**1. Why divide attention scores by √d?**

Dividing by the square root of the key dimension prevents dot-product scores from becoming excessively large. This helps keep the softmax distribution from becoming too peaked and supports more stable gradients during training.

**2. How does self-attention help a model understand relationships between words?**

Self-attention allows each token to assign different importance weights to other tokens in the same sequence. It can connect words that are far apart and use their context to build a more informative representation of each word.
